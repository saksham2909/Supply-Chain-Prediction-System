```python
# dataset load kr rhe h

import pandas as pd

df = pd.read_csv(
    "../data/DataCoSupplyChainDataset.csv",
    encoding="latin-1"
)

df.head()
```

```python
# columns check kr rhe h

print("Total columns:", len(df.columns))
print(df.columns.tolist())
```

```python
# dataset ka size dekh rhe h

print("Shape:", df.shape)

# dataset ki information dekh rhe h

df.info()

# numerical columns ka basic summary

df.describe()
```

```python
# missing values ka percentage check kr rhe h

missing = df.isnull().sum()
missing_percent = (missing / len(df)) * 100

missing_percent[missing_percent > 0].sort_values(ascending=False)
```

```python
# pura empty column remove kr rhe h

df = df.drop(columns=["Product Description"])
```

```python
# bahut zyada missing values wala column remove kr rhe h

df = df.drop(columns=["Order Zipcode"])
```

```python
# customer last name ki missing values fill kr rhe h

df["Customer Lname"] = df["Customer Lname"].fillna(
    df["Customer Lname"].mode()[0]
)
```

```python
# customer zipcode ki missing values fill kr rhe h

df["Customer Zipcode"] = df["Customer Zipcode"].fillna(
    df["Customer Zipcode"].median()
)
```

```python
# dobara missing values check kr rhe h

df.isnull().sum()[df.isnull().sum() > 0]
```

```python
# duplicate rows check kr rhe h

df.duplicated().sum()
```

```python
# saare columns dekh rhe h

df.columns.tolist()
```

```python
# personal aur useless columns remove kr rhe h

df = df.drop(columns=[
    "Customer Email",
    "Customer Fname",
    "Customer Lname",
    "Customer Password",
    "Customer Street",
    "Product Image"
])
```

```python
# columns kitne bache check kr rhe h

print("Columns:", len(df.columns))
```

```python
# date columns ko datetime me convert kr rhe h

df["order date (DateOrders)"] = pd.to_datetime(
    df["order date (DateOrders)"]
)

df["shipping date (DateOrders)"] = pd.to_datetime(
    df["shipping date (DateOrders)"]
)
```

```python
# date ka datatype check kr rhe h

print(df["order date (DateOrders)"].dtype)
print(df["shipping date (DateOrders)"].dtype)
```

```python
# target ki values check kr rhe h

df["Late_delivery_risk"].value_counts()
```

```python
# target ka percentage check kr rhe h

df["Late_delivery_risk"].value_counts(normalize=True) * 100
```

```python
# actual aur scheduled shipping ka difference nikal rhe h

df["Delay_Days"] = (
    df["Days for shipping (real)"]
    - df["Days for shipment (scheduled)"]
)

df[["Days for shipping (real)", "Days for shipment (scheduled)", "Delay_Days"]].head()
```

```python
# delay ko categories me divide kr rhe h

df["Delay_Category"] = pd.cut(
    df["Delay_Days"],
    bins=[-100, 0, 2, 100],
    labels=["Early_On_Time", "Small_Delay", "High_Delay"]
)

df["Delay_Category"].value_counts()
```

```python
# months check kr rhe h
print(df.columns.tolist())
```

```python
# Step 1: date column ko datetime type me convert karo
df['order date (DateOrders)'] = pd.to_datetime(df['order date (DateOrders)'])

# Step 2: ab Month column bana sakte ho
df['Month'] = df['order date (DateOrders)'].dt.month

# ab yeh chalega
df["Month"].value_counts()
```

```python
# top 3 months by order volume ko peak season maan rahe hain
peak_months = [1, 5, 7]
df['Peak_Season'] = df['Month'].apply(lambda x: 1 if x in peak_months else 0)
```

```python
# 1. Peak season flag (based on actual top months)
peak_months = [1, 5, 7]
df['Peak_Season'] = df['Month'].apply(lambda x: 1 if x in peak_months else 0)

# 2. Weekend flag
df['Order_Day'] = df['order date (DateOrders)'].dt.dayofweek
df['Is_Weekend'] = df['Order_Day'].apply(lambda x: 1 if x >= 5 else 0)

# 3. Shipping delay gap (scheduled vs real)
df['Shipping_Gap'] = df['Days for shipping (real)'] - df['Days for shipment (scheduled)']

# 4. Order value category (Sales column ko bins me daalo)
df['Order_Value_Category'] = pd.cut(df['Sales'], bins=[0,100,300,1000,float('inf')],
                                     labels=['Low','Medium','High','Very High'])

# 5. Quick insight check - region wise late delivery rate
df.groupby('Order Region')['Late_delivery_risk'].mean().sort_values(ascending=False)
```

```python
# 1. Check final missing values
print(df.isnull().sum().sort_values(ascending=False).head(10))

# 2. Drop irrelevant/redundant columns (adjust list based on your data)
drop_cols = ['Customer Zipcode', 'Product Status', 'Order Status']  # example — check what's actually useless
df.drop(columns=drop_cols, inplace=True, errors='ignore')

# 3. Final shape check
print(df.shape)

# 4. Export cleaned dataset
df.to_csv('cleaned_supply_chain.csv', index=False)
print("Saved successfully")
```
